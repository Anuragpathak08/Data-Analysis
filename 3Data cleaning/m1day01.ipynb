## What is Data Cleaning? <hr>
Data cleaning is the process of identifying and correcting (or removing) errors, inconsistencies, and inaccuracies in a dataset to improve its quality and reliability.

It involves transforming raw, messy data into a clean, structured format that's ready for analysis, visualization, or machine learning.



**The 6 Steps of Data Cleaning**

#### **1. Renaming the Columns**<hr>
* Changing column names to make them more descriptive, consistent, and easier to understand.

* A clear column names help anyone reading the dataset understand what each variable represents without confusion. 
    
* It also makes coding easier and reduces errors when referring to columns in analysis.

````python
# In order to perform the data cleaning i will use Titanic Data set in this data set three col name are not very clear sibsp, parch, fare
rename = {'SibSp':'Sibling/Spouse',
          'Parch': 'Parent/Child',
          'Fare':'Ticket_fare'}

df.rename(columns=rename, inplace=True)
df.head(1)
````

#### **2. Removing Duplicate Rows**<hr>
* Identifying and deleting identical rows that appear multiple times in your dataset.

* Duplicate entries can artificially inflate your data, leading to incorrect analysis results and biased insights. Removing them ensures each observation is counted only once.

`since there is no Dupicated data we are not removing anything`

````python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
````

````python
df.duplicated(subset=['Ticket','Name']).sum()
df.drop_duplicates(subset=['Ticket','Name'], inplace=True)
````

#### **3. Changing the Data Present in a Particular Column**<hr>
*  Standardizing inconsistent values within a column (like "First", "1st", "FIRST" all becoming "1st") or correcting incorrect entries.

* Inconsistent categorical data can break analysis and cause errors in grouping, filtering, and visualization. Uniform values ensure accurate aggregation and comparison.

syntax to chage values in a column : 
```
    d = {old_value : new_value, ....}
    df[COLUMN] = df[COLUMN].map( d )
```

If we want to change one value we need to mention all the unique value.

````python
df['Pclass'].unique()
````

````python
change_values = {'1':'1st',
                 '2':'2nd',
                 '3':'3rd',
                 '3rd':'3rd',
                 '1st':'1st'}
df['Pclass'] = df['Pclass'].map(change_values)
````

````python
df.sample(2)
````

#### **4. Handling NaN Values**<hr>
* Dealing with missing data points (Not a Number, None, or null values) by either removing rows with missing data or filling them with appropriate substitute values.

* Many algorithms (Data analysis or Data science algorithms) cannot process missing values, and they can skew statistical analysis. 

* The choice between dropping or filling depends on how much data is missing and its importance to your analysis.

````python
# we can handle the nan in embarked column

df['Embarked'].isnull().sum() # since there is only 3 null/nan/None values we can remove it.

df.dropna(subset=['Embarked'], inplace=True)
````

````python
df['Embarked'].isnull().sum()
````

````python
# we can handle the nan in age column
df['Age'].isnull().sum()

age_median = df['Age'].median()
df['Age'] = df['Age'].fillna(age_median)
````

````python
df['Age'] = df.Age.astype('int32')
df.Age
````

#### **5. Dropping Unwanted Columns**<hr>
* Removing entire columns that are irrelevant, redundant, or contain too much missing information to be useful.

* Extra unnecessary columns add noise, increase data size, and slow down processing.

* Focusing on relevant features improves model performance and makes analysis more efficient.

````python
df.drop(columns=['Cabin'], inplace=True)
````

````python
df.head(1)
````

#### **6. Dropping Unwanted Rows**<hr>
* Removing specific rows from the dataset that contain outliers, errors, or irrelevant observations.

* Outliers or erroneous entries can distort statistical measures and negatively impact model training. 

* Removing them ensures your analysis is based on accurate, representative data.

````python
df.drop(index=[887], inplace=True)
````
