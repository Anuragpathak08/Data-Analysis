## Indexing on Pandas Series
`sytnax : variable[index value]`

```python
s = pd.Series([1,23,451,23,456])
s[0] # np.int64(1)
s[1] # np.int64(23)
s[2] # np.int64(451)
s[2] = 10 # items Assignment is also possible 
s
# s[9] # keyerror because index element not exist
```

```python
# s[-1] #keyerror because index element dosent exist
```

```python
marks = {'physics': 83,
         'chemistry': 89,
         'maths': 84,
         'bio': 80,
         'English': 70,
         'Hindi': 70}

s2 = pd.Series(marks)
s2
```

```python
# s2[0] # Error because we can only perform indexing with the label
s2['physics']
```

## Slicing on Pandas Series<hr>
`syntax : variable[start: end: step]`

```python
s1 = pd.Series(np.arange(1, 20, 1))
s1[0:10:1]
s1[0:20:2]
s1[10::-1]
s1[ :10:-1]
s1[1:-10:2]
```

```python
s2['physics': 'Hindi': 1]
s2[0::1]
```

```python
s3 = pd.Series(np.arange(7), index = [0,10,20,30,40,50,60])
s3[0:10:1]
```

```python
s3[0:7:1]
s3[0:10:1]
```

## Boolean Filtering on Pandas Series<hr>
`syntax : variable[condition]`

```python
s1[s1%2==0]
# s1[(s1%2==0) or (s1%3==0)] # error coz logical is not allowed
# s1[s1 == 10 | 20== s1 ] #error no bracket
s1[(s1 == 10) | (20 == s1) ]
s1[(s1 % 5== 0) & (s1%3 != 0)]
```

```python
s4 = pd.Series(np.random.randint(0, 2, 10))
```

```python
s4[~(s4==1)]
```
