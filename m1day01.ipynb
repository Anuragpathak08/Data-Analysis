**syntax to create a numpy array** 
`np.array(value)`

```python
arr2 = np.array([10,20,30,40,50,60])
arr2
```

**int, float, complex, bool, str** : These are  values, When passed to an array, they create a 0-dimensional array containing just that one value.

**set, dict** : NumPy does not unpack these. It treats the entire dictionary or set as a single "object".

```python
np.array(10)
np.array(10.234)
np.array(20+3j)
np.array(True)
np.array('hello')
np.array({1,3,334546.46,5,7,567})
np.array({1:2, 3:4})
```

**list, tuple** : These are naturally unpacked by NumPy to create a standard, multi-dimension array (e.g., 1D, 2D arrays)

```python
np.array([1,2,3,4,56])
np.array((1,2,3,4,56))
np.array(['hi', 'hello'])
```

```python
np.array([1,2,3,4,{1}])
# np.array(['hello', {1}, ()]) #error 
```

If you create a NumPy array with heterogeneous (mixed) data types, NumPy automatically converts all elements into a homogeneous data type. 
It determines the final data type by following a strict upcasting hierarchy (priority list):

- bool (Lowest priority)
- int
- float
- complex
- str (String)
- object (set, dict, custom classes) (Highest priority)

```python
np.array([True, False,])
```

```python
np.array([1,2,3, True, False])
```

```python
np.array([1, 2.234, 12,  13.2, True, False])
```

```python
np.array([1, 2, 12,  (13+ 2j) , True, False])
```

```python
np.array(['hello',1,2,4, 10j, 34.45])
```

```python
np.array(['hello',1,2,4, 10j, 34.45, {122:4}, {1,23}])
```

# Types of arrays:
1) zero-dimensional array (0D)

```python
np.array(10) #0D array
```

2) Uni-dimensional array (1D)

```python
np.array((1,23,4,34,23,46,57,6,43)) #1D array
np.array([123,4,4,6,75]) #1D array
```

3) Bi-dimensional array (2D)

```python
np.array([[1,2,4],
          [3,5,45]]) #2D array
```

```python
# np.array([[1,2,3],
#           [1,23]]) # ValueError :- because the size of first element is not same as second
```

4) Multi-dimensional array (nD)

```python
np.array([[[ 1,  2,  3],
           [ 1, 23, 34]],
           
          [[ 1,  2,  3],
           [ 1, 23, 34]]])
```
