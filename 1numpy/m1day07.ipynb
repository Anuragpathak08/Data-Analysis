**Copy** → modification will not affect both arrays.<br>
`syntax : arr.copy()`

```python
arrx = arr1.copy()
arrx[2] = 300
arrx
```

```python
arr1
```

```python
arr2x = arr2.copy()
arr2x[1][1]  = 2000
arr2x
```

```python
arr2
```

```python
print(id(arr2))
print(id(arr2x))
```

**view** → if modification will affect both arrays but address will be different<br>
`syntax : arr.view()`

```python
arrx = arr1.view()
print(id(arrx))
print(id(arr1))
```
