# Attributes of array 

`array.ndim` : Returns the number of dimensions (axes) of the array.

```python
arr1 = np.array([1,23,45])
arr1.ndim
```

```python
arr2 = np.array([[1,23,4],
                 [1,2,3]])
arr2.ndim
```

`array.shape` : Returns a tuple showing the size of the array along each dimension (e.g., rows and columns)

```python
arr3 = np.array([1,2,3])
arr3.shape
```

```python
arr4 = np.array([[1,2,3],
                 [4,5,67]])
arr4.shape
```

```python
arr5 = np.array([[[12,34,34],
                 [2,3,4],
                 [1,2,3]],
                 
                [[2,3,4],
                 [1,2,34],
                 [4,5,67]]])
arr5.shape
```

`array.dtype` : Returns the data type of the array's elements (e.g., int64, float64).

```python
arr1.dtype
```

```python
arr6 = np.array([[12.0 ,1,2,4],
                 [2,4,5,7]])
arr6.dtype
```

```python
arr6 = np.array([['hello' ,1,2,4],
                 [2,4,5,7]])
arr6.dtype
```

```python
arr7 = np.array(['hello'])
arr7.dtype
```

`array.itemsize` : Returns the size in bytes of each individual array element.

```python
arr = np.array([199, 3666, 34,'hello'])
arr.itemsize
```

`array.nbytes` : Returns the total number of bytes consumed by all elements in the array.

```python
arr.nbytes
```

# Indexing 

<hr>

**Indxeing of 1D array**
`syntax : variable[index value]`

```python
arr
```

```python
arr[0]
arr[-1]
arr[-3]
```

<hr>

**Indxeing of 2D array**
`syntax : variable[row][column]`

```python
arr2 = np.array([[1,3,45],
                 [13,45,6],
                 [13,4,56]])
```

```python
arr2[0][0]
# arr2[1][3] # Error :- index out of bound  
arr2[2][-1]
```

<hr>

**Indxeing of 3D array**
`syntax : variable[depth][row][column]`

```python
arr3 = np.array([ [ [12, 34, 34],
                    [ 456,5,75],
                    [ 234, 34,45]],

                  [ [ 2,  3,  434],
                    [ 13, 42, 84],
                    [ 4,  5, 67]]])
```

```python
arr3[0][1][2]
arr3[-1][2][-2]
arr3[-1, 2, -2]
```
