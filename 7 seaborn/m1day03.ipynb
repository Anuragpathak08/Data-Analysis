````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

### **Matrix Plot**
A matrix plot is a visualization where numerical values are arranged in rows and columns, and each cell of the matrix is represented visually.
```example
       A   B   C
    ┌─────────────
X   │ 10  20  30
Y   │ 40  50  60
Z   │ 70  80  90
```

#### **Heatmap**
- A heatmap represents matrix values using different colors.

```python
syntax:- sns.heatmap(data=glue, annot=True, fmt='.1f', cmap='crest',
                     linewidths=1, linecolor='white')
```

##### Parameters

- `data` → The data you want to represent as a heatmap, usually a DataFrame or matrix.
- `annot=True` → Displays the actual values inside each cell.
- `fmt='.1f'` → Formats the displayed values to **1 decimal place**.
- `cmap='crest'` → Defines the color palette used for the heatmap.
- `linewidths=1` → Sets the width of the lines between cells.
- `linecolor='white'` → Sets the color of the lines between cells.


````python
d = {'maths':[89,67, 90],
     'english':[67,56,76],}
df = pd.DataFrame(d, index= ['a','b','c'])

sns.heatmap(df)
df
````

````python
glue = sns.load_dataset("glue").pivot(index="Model", columns="Task", values="Score")
glue
````

````python
sns.heatmap(glue)
````

````python
sns.heatmap(data = glue, vmin= 10 ,vmax=100)
````

````python
sns.heatmap(data = glue, annot=True, fmt='.1f',cmap='crest', linewidths=1, linecolor='white')
plt.xticks(rotation = 90)
````
