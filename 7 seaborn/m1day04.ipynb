##### Figure level function -> `catplot`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic')
titanic.info()
```

#### Figure- level function 
 `syntax : catplot(kind, data, x, y, jitter, hue, col, row)`

```python
sns.catplot(kind= 'strip' ,data=titanic, x = 'age', y = 'class',hue='survived', jitter = 0.2)
```

```python
sns.catplot(kind= 'strip' ,data=titanic, x = 'age', y = 'class', col= 'sex', row='alive')
```

```python
tips = sns.load_dataset('tips')
tips.head(3)
```

#### Axes-level 
`syntax : stripplot(data, x, y, hue)`

```python
sns.stripplot(data=tips, x= 'day', y = 'total_bill', hue='day')
```

#### Swarm plot

```python
sns.catplot(kind='swarm', data=tips, x = 'day', y= 'total_bill', hue='sex')
```

Axes-level function `syntax: sns.swarnplot(data,x,y,hue)`

```python
sns.swarmplot(data=titanic, x = 'class', y = 'age')
```

### **Boxplot**

A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

image.png

```python
sns.catplot(kind='box' ,data=tips, x = 'total_bill')
```

```python
sns.boxplot(data=tips, x = 'tip', y= 'day',)
```

```python
sns.boxplot(data=titanic, x = 'age', y = 'class')
```

```python
sns.boxplot(data=titanic, x = 'age', y = 'class', hue='survived')
```
