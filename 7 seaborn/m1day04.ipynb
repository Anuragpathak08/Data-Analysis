```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic')
titanic.info()
```

#### Figure- level function 
 `syntax : catplot(kind, data, x, y, hue, col, row)`

```python
sns.catplot(kind= 'strip' ,data=titanic, x = 'age', y = 'class',hue='survived')
```

```python
sns.catplot(kind= 'strip' ,data=titanic, x = 'age', y = 'class', col= 'sex', row='alive')
```

```python
tips = sns.load_dataset('tips')
tips.head(3)
```

#### Axes-level 
`syntax : stripplot(data, x, y, hue)`

```python
sns.stripplot(data=tips, x= 'day', y = 'total_bill', hue='day')
```
