````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
tips = sns.load_dataset('tips')
penguins = sns.load_dataset('penguins')
flight = sns.load_dataset('flights')
titanic = sns.load_dataset('titanic')
````

### **Multi-plot**
- Facet plot
- Join plot
- Pair plot 

##### **Facet plot**
- Facet plot using a figure level function

````python
# sns.relplot(tips, x = 'total_bill', y = 'tip', col = 'day', col_wrap=2)
sns.relplot(tips, x = 'total_bill', y = 'tip', col = 'day', row= 'time', height=3 ,aspect=1.2)
````

- Facet plot using Facet Function
*syntax*
```python 
var = sns.FacetGrid(data, col = col_name, row= row_name)
var.map(plot_name, x, y)
```
---

````python
s = sns.FacetGrid(data=tips, col='day', row='time')
s.map(sns.scatterplot,  'tip', 'total_bill')
````

*syntax*
```python 
var = sns.FacetGrid(data, col = col_name, row= row_name)
var.map_dataframe(plot_name, x= 'x-axis', y = 'y-axis')
```
---

````python
s = sns.FacetGrid(data=tips, row = 'sex', col = 'day')
s.map_dataframe(sns.histplot, x = 'total_bill', hue = 'time')
s.add_legend()
````

### Pairplot

````python
sns.pairplot(tips, hue='sex')
````

````python
sns.pairplot(data=titanic, diag_kind='kde', kind='hist', corner=True)
````

### PairGrid

````python
p = sns.PairGrid(penguins, hue='species')
p.map(sns.scatterplot)
p.add_legend()
````

````python
p = sns.PairGrid(penguins, hue='species')
p.map_diag(sns.histplot)
p.map_lower(sns.kdeplot)
p.map_upper(sns.scatterplot)
p.add_legend()
````
