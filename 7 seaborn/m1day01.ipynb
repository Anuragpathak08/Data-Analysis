# Seaborn
Seaborn is a popular Python data visualization library built on top of Matplotlib that makes it easy to create attractive and informative statistical graphics.

### Why Seaborn?
- provides a layer of abstraction hence simpler to use
- better aesthetics
- more graphs included

### Types of Functions
- Figure Level
- Axis Level

1.  Axes-level functions: Plot data onto a single Matplotlib Axes object. Examples include sns.scatterplot() or sns.histplot().
2. Figure-level functions: Manage the entire figure using a grid (like a FacetGrid). They use a kind parameter to switch between underlying plots. Examples include sns.relplot(), sns.displot(), and sns.catplot().

Main Classification of Seaborn plots

- Relational Plot
- Distribution Plot
- Categorical Plot
- Regression Plot
- Matrix Plot
- Multiplots

https://seaborn.pydata.org/api.html

### 1. Relational Plot
A Seaborn relational plot is a high-level figure-level interface—accessed via `sns.relplot()` used to visualize the statistical relationships between numerical variables

- To see the statistical relation between 2 or more variables.
- Bivariate Analysis

Plots under this section

- scatterplot
- lineplot

`syntax : sns.relplot(data, x, y, kind )` 

#### parameters of relplot 
- kind: Specifies the style of the plot; set to "scatter" (default) or "line".
- x, y: Names of variables in data or vector data to map to the x and y axes.
- data: Pandas DataFrame or array-like object where the variables exist.
- hue: Grouping variable that maps distinct colors to data points based on a categorical or numerical column.
- size: Grouping variable that changes the size of the markers (in scatter) or line width (in line plots) based on values.
- style: Grouping variable that changes the marker shape or line dash pattern (e.g., solid vs. dashed).col, 
- row: Categorical variables that split the data into separate columns or rows of subplots (facets).
- col_wrap: Maximum number of column facets to display before wrapping into the next row.

```python
import pandas as pd
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
tips.head(2)
```

```python
# figure level function
sns.relplot(kind = 'scatter',
            data = tips, 
            x = 'tip', 
            y = 'total_bill' , 
            hue = 'sex',
            size = 'size',
            style='smoker')
```

```python
# Axis level function
sns.scatterplot(data = tips, 
                x = 'tip', 
                y = 'total_bill',
                hue = 'sex',
                size = 'size',
                style = 'smoker' )
```

```python
gap =  pd.read_csv('gapminder.csv')
temp_df = gap[gap['country'].isin(['Afghanistan','Brazil', 'Pakistan', 'India'])]
```

```python
# figure level
sns.relplot(kind='line',
            data=temp_df,
            x = 'year',
            y = 'gdp',
            hue ='country',
            style='continent')
```
