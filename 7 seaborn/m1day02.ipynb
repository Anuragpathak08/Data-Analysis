```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
penguins = sns.load_dataset('penguins')
```

### 2. Distribution Plots
A distribution plot is a graphical tool used to display how data values are spread out and how often each value appears.

- It is used for univariate analysis
- It is used to find out the distribution
- It shows Range of the observation
- It is also use to find Central Tendency
- Is the data bimodal?
- Are there outliers in the data?

Plots under distribution plot

- histplot
- kdeplot
- rugplot

Figure-level function for showing data distributions in Seaborn is `sns.displot()`

Axes-level function for showing data distributions in Seaborn
* `sns.histplot()`: Plots a univariate or bivariate histogram to show the count of observations within discrete bins.
* `sns.kdeplot()`: Computes a Kernel Density Estimate to display a smooth, continuous probability density curve.
* `sns.rugplot()`: Draws small vertical ticks along an axis to represent every individual observation.

#### **Histogram** 
figure level function - `syntax : sns.displot(kind = 'hist', data, x, bin, hue)`

```python
# sns.displot(kind='hist', data=tips, x = 'tip')
# sns.displot(kind='hist', data=tips, x = 'tip', hue='sex')
sns.displot(kind='hist', data=tips, x = 'tip', bins = 18)
```

Axes level function -`syntax : sns.histplot( data, x, hue)`

```python
sns.histplot(data=penguins, x = 'flipper_length_mm' ,bins= 20, hue='sex')
```

```python
sns.displot(kind = 'hist' ,data=penguins, x = 'flipper_length_mm' ,bins= 20, col = 'sex' )
```

#### **Bivariate Histogram** 

```python
sns.displot(kind='hist', data=penguins, x='bill_length_mm', y='bill_depth_mm', bins= 18,color='red')
```

```python
sns.histplot(data=tips, x = 'tip', y= 'total_bill', )
```

#### **KDE plot**
figure level function - `syntax : sns.displot(kind = 'kde', data, x, bin, hue)`

```python
sns.displot(kind='kde',data=penguins, x='flipper_length_mm')
```

Axes-level function - `syntax : sns.displot(kind = 'kde', data, x, bin, hue)`

```python
sns.kdeplot(data=tips, x = 'tip')
```

#### **Bivariate KDE**

```python
sns.displot(kind = 'kde', data= tips, x = 'tip', y = 'total_bill', hue='sex', palette=['red', 'green'])
# sns.displot(kind = 'kde', data= tips, x = 'tip', y = 'total_bill', color = 'red' )
```

#### **Histogram with KDE**

```python
sns.displot(kind='hist', data=penguins, x = 'flipper_length_mm' ,kde = True)
```

#### **Rugplot**
This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
- Plot marginal distributions by drawing ticks along the x and y axes.

```python
sns.kdeplot(data=tips, x = 'tip')
sns.rugplot(data=tips, x = 'tip')
```

```python
sns.displot(kind='kde',data=tips, x= 'tip',rug=True)
```
