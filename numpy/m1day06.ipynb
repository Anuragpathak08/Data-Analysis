## Methods of array <hr>

**reshape** → To change the dimension of array. <br>
`syntax : array.reshape(row, column)`

```python
arr1.reshape(2,10)
# arr1.reshape(2,15) # Error : the product of row and column is not equals to the number
                             # of elements
```

3d array
`syntax:array.reshape(depth,row, column)`

```python
arr2.reshape(2,8)
# arr2 # original didnt get modified 
# arr2.reshape(1.6, 10 ) #error cannot be a float
arr2.reshape(2,2,4)
arr2.reshape(16)
```

**arange** → Used to create array between specified range. We can generate float values also. <br>
`syntax : np.arange(start, end+1, step)`

```python
arr1 = np.arange(20, 100, 3)
arr1
```

```python
arr2 = np.arange(0,9).reshape(3,3)
arr2
```

<p>arange with float</p>

```python
arr3 = np.arange(0,10.23, 0.5)
arr3
```

**linspace** → Create an array with equally spaced values. By default create an array with 50 values. <br>
`Syntax: np.linspace(start, end, number of values)`

```python
np.linspace(1, 100, 9)
```

```python
np.linspace(1, 50)
np.linspace(1, 50).reshape(5, 10)
```

**zeros** → Used to create array with zeros. <br>
`Syntax: 1D → np.zeros(n)`
`2D → np.zeros((r, c))`

```python
arr1 = np.zeros(5)
arr1
```

```python
arr2 = np.zeros((2,5))
arr2
```

```python
arr3 = np.zeros((3,3,3))
arr3
```

**ones** → Used to create array filled with ones.<br>
`syntax: 1D → np.ones(n)`
`2D → np.ones((r, c))`


```python
np.ones(10)
```

```python
np.ones((2,3))
```

```python
np.ones((2, 2, 2))
```

**Identity**→ used to create identity matrix<br>
Rules of identity matrix: rows = columns , primary diagonal should be 1 <br>
`syntax : np.identity(n)` 



```python
np.identity(3)
```

```python
arr1 = np.identity(3, dtype=int)
arr1
```

```python
arr1.dtype
```

**random.randint** → It is used to create array with random values within the
specified limit. It will only generate integer values. <br>
`syntax: np.random.randint(start, end+1, no of values)`

```python
np.random.randint(1, 100, 10)
```

```python
np.random.randint(2,1000, 3000)
```

```python
np.random.randint(2,10, 6).reshape((2,3))
```

**random.rand** → It is used to create an array with positive values b/w 0 to 1 only (exclusive).<br>
`syntax:np.random.rand(n)`

```python
np.random.rand()
```

```python
np.random.rand(2)
```

```python
np.random.rand(10) * 100
```
