```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### Bar plot (bar graph)

```python
subject = ['English', 'Hindi', 'Marathi', 'Science', 'Maths', 'SS']
plt.subplot(2,2,1)
_marks = [90, 87, 89, 9, 100, 78.09]
plt.bar(subject, _marks, color='#F77E2D', width=0.4, label = 'Modi')
plt.legend()

plt.subplot(2,2,2)
rahul_marks = [100, 60, 60, 70, 76, 99 ]
plt.bar(subject, rahul_marks, color = '#5469c7', width=0.4, label = 'rahul')
plt.legend()
```

```python
plt.bar([0.2,1.2, 2.2, 3.2, 4.2, 5.2], _marks, width=0.4)
plt.bar([-0.2,0.8, 1.8, 2.8, 3.8, 4.8], rahul_marks, width=0.4)

```

```python
x = np.arange(len(subject))
width = 0.4

plt.bar(x - width/2, _marks, width=width, color='#F77E2D', label='Modi')
plt.bar(x + width/2, rahul_marks, width=width, color='#5469c7', label='Rahul')

# plt.xticks(x, subject)
plt.legend()
plt.show()
```
