# Data Visualization using Matplotlib

Data visualization is an important part of **data analysis and interpretation**. It represents data in a graphical format, making it easier to understand **trends, patterns, comparisons, and relationships** within the data.

In Python, **Matplotlib** is one of the most widely used libraries for creating **static, animated, and interactive visualizations**.

## Introduction to Matplotlib

**Matplotlib** is a comprehensive Python library used to create various types of visualizations and plots.

The most commonly used module in Matplotlib is **`pyplot`**, which provides a simple interface for creating and customizing plots.

### Importing Matplotlib

```python
import matplotlib.pyplot as plt
````

Here, `plt` is an **alias** used to access the functions provided by the `pyplot` module conveniently.

---

## Basic Plotting with Matplotlib

The simplest visualization that can be created using Matplotlib is a **line plot**.

The `plot()` function is used to create a line plot.

### Syntax

```python
plt.plot(x, y)
```

Where:

* `x` → Represents the data points on the **x-axis**.
* `y` → Represents the data points on the **y-axis**.

### Example

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.plot(x, y)
plt.show()
```

---

# Important Attributes of `plot()`

The appearance of a plot can be customized using various attributes of the `plot()` function.

| Attribute         | Description                                                        | Example                                                |
| ----------------- | ------------------------------------------------------------------ | ------------------------------------------------------ |
| `color`           | Specifies the color of the line.                                   | `plt.plot(x, y, color='red')`                          |
| `label`           | Assigns a name to the line, which can be displayed in the legend.  | `plt.plot(x, y, label='Line 1')`                       |
| `linestyle`       | Specifies the style of the line, such as solid, dashed, or dotted. | `plt.plot(x, y, linestyle='--')`                       |
| `linewidth`       | Specifies the thickness of the line.                               | `plt.plot(x, y, linewidth=3)`                          |
| `marker`          | Specifies the symbol used to represent individual data points.     | `plt.plot(x, y, marker='o')`                           |
| `markeredgecolor` | Specifies the color of the marker's boundary.                      | `plt.plot(x, y, marker='o', markeredgecolor='black')`  |
| `markerfacecolor` | Specifies the color inside the marker.                             | `plt.plot(x, y, marker='o', markerfacecolor='yellow')` |

## Common Marker Examples

Some commonly used markers are:

* `o` → Circle
* `^` → Triangle
* `s` → Square
* `*` → Star
* `+` → Plus
* `x` → X-shaped marker

### Example with Multiple Attributes

```python
plt.plot(
    x, y,
    color='red',
    label='Sales',
    linestyle='--',
    linewidth=2,
    marker='o',
    markeredgecolor='black',
    markerfacecolor='yellow'
)
plt.show()

```

`````python

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.plot(x, y)
plt.show()
`````

### Plotting Multiple Lines

- Matplotlib allows us to plot **multiple lines on a single graph**.
- Each line can represent a different set of data.
- Multiple `plt.plot()` statements can be used to add different lines to the same plot.
- The `label` attribute can be used to identify each line in the legend.

### Example

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

y1 = [10, 20, 15, 25, 30]
y2 = [15, 25, 20, 30, 35]

plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")

plt.show()
```

`````python
x = [1, 2, 3, 4, 5]

y1 = [10, 20, 15, 25, 30]
y2 = [15, 25, 20, 30, 35]

plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")

plt.show()
`````

## Labelling and Titling the Graph

To make a graph clear and meaningful, it is important to **label the axes** and provide an appropriate **title**. Matplotlib provides separate functions for this purpose.

| Function | Description | Example |
|---|---|---|
| `xlabel()` | Adds a label to the x-axis. | `plt.xlabel("Time (seconds)")` |
| `ylabel()` | Adds a label to the y-axis. | `plt.ylabel("Speed (m/s)")` |
| `title()` | Adds a title to the graph. | `plt.title("Speed vs Time")` |

### Example

```python
import matplotlib.pyplot as plt

time = [1, 2, 3, 4, 5]
speed = [10, 20, 30, 40, 50]

plt.plot(time, speed)

plt.xlabel("Time (seconds)")
plt.ylabel("Speed (m/s)")
plt.title("Speed vs Time")

plt.show()

`````python
plt.plot([1, 2.20, 4.4, 6.8], [0, 10, 40, 35] , marker='o')
plt.xlabel('time')
plt.ylabel('speed')
plt.title('speed v/s time')
plt.show()
`````

## Legend in a Graph

A **legend** provides information about the different lines or data series displayed on a graph. It helps the viewer identify what each line represents.

The `label` attribute is used to assign a name to each line, and the `legend()` function is used to display those labels on the graph.

### Example

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.plot(x, y, label="Line 1")
plt.plot(x, [i**2 for i in y], label="Line 2")

plt.legend()
plt.show()
````

* `label` → Assigns a name to each line.
* `plt.legend()` → Displays the labels as a legend on the graph.

---

## Customizing the Legend Location

The position of the legend can be customized using the `loc` attribute.

### Syntax

```python
plt.legend(loc="upper right")
```

### Common Legend Locations

| Location         | Description                                   |
| ---------------- | --------------------------------------------- |
| `"upper left"`   | Places the legend in the upper-left corner.   |
| `"upper right"`  | Places the legend in the upper-right corner.  |
| `"lower left"`   | Places the legend in the lower-left corner.   |
| `"lower right"`  | Places the legend in the lower-right corner.  |
| `"center"`       | Places the legend at the center of the graph. |
| `"center left"`  | Places the legend at the center-left.         |
| `"center right"` | Places the legend at the center-right.        |
| `"lower center"` | Places the legend at the lower-center.        |
| `"upper center"` | Places the legend at the upper-center.        |



`````python
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.plot(x, y, label="Line 1")
plt.plot(x, [i**2 for i in y], label="Line 2")

plt.legend()
plt.show()
`````

## Ticks on Axes

**Ticks** are marks displayed along the x-axis and y-axis that help identify specific values or positions on a graph.

Matplotlib allows us to customize the **position and labels of ticks** using the `xticks()` and `yticks()` functions.
### Syntax
```python
plt.xticks([0, 1, 2, 3], ['A', 'B', 'C', 'D'])
plt.yticks([10, 20, 30], ['Low', 'Medium', 'High'])
```

### `xticks()`
The `xticks()` function is used to customize the **x-axis tick positions and labels**.
```python
plt.xticks([0, 1, 2, 3], ['A', 'B', 'C', 'D'])
```

* `[0, 1, 2, 3]` → Specifies the positions of the ticks.
* `['A', 'B', 'C', 'D']` → Specifies the labels displayed at those positions.

### `yticks()`
The `yticks()` function is used to customize the **y-axis tick positions and labels**.

```python
plt.yticks([10, 20, 30], ['Low', 'Medium', 'High'])
```

* `[10, 20, 30]` → Specifies the positions of the ticks.
* `['Low', 'Medium', 'High']` → Specifies the labels displayed at those positions.

### Example

```python
import matplotlib.pyplot as plt

x = [0, 1, 2, 3]
y = [10, 20, 30, 20]

plt.plot(x, y)

plt.xticks([0, 1, 2, 3], ['A', 'B', 'C', 'D'])
plt.yticks([10, 20, 30], ['Low', 'Medium', 'High'])

plt.show()
```

`````python
x = [0, 1, 2, 3]
y = [10, 20, 30, 20]

plt.plot(x, y)

plt.xticks([0, 1, 2, 3], ['A', 'B', 'C', 'D'])
plt.yticks([10, 20, 30], ['Low', 'Medium', 'High'])

plt.show()
`````
