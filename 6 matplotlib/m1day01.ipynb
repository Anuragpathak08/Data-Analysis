```python
nashedi = pd.read_csv('nashedi_student.csv')
nashedi_2 = nashedi.groupby('gender')
plt.plot(nashedi_2.get_group('F')['Dalc'].unique(), label = 'Female nashedis')
plt.plot(nashedi_2.get_group('F')['Walc'].unique(), label = 'Female nashedis on sunday')
plt.plot(nashedi_2.get_group('M')['Dalc'].unique(), label = 'Male nashedis')
plt.plot(nashedi_2.get_group('M')['Walc'].unique(), label = 'Male nashedis on sunday')
plt.legend(loc = 'upper left')
plt.yticks([1, 3, 5], ['Low', 'Medium', 'High'])
```

```python
imdb = pd.read_csv(r'..\4 groupby\imdb_top_1000.csv')
acd = imdb.groupby('Genre').get_group('Action, Crime, Drama')['Gross'].dropna()
asf = imdb.groupby('Genre').get_group('Action, Sci-Fi')['Gross'].dropna()
west = imdb.groupby('Genre').get_group('Western')['Gross'].dropna()
# imdb['Genre'].unique()
```

```python
acd = acd.str.replace(',', '').astype('int32').tail()
asf = asf.str.replace(',', '').astype('int32').tail()
west = west.str.replace(',', '').astype('int32').tail()
```
