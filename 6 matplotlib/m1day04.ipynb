```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Histogram

```python
ages = np.random.randint(16, 70, 500)
```

```python
plt.hist(ages, bins=10, edgecolor = 'black')
```

```python
bins = [10,20,30,40,50,60,70,80]
# bins =list( range(15,75, 10))
plt.hist(ages, bins= bins, edgecolor = 'black')
```

```python
bins = range(15,75,5)
plt.hist(ages, bins= bins, edgecolor = 'black', linewidth = 0.5, label='Dev ages')
plt.xlabel('Age Groups')
plt.ylabel('Count')
plt.title('Age Distribution of Employees')
plt.legend()
plt.show()
```

## Scatter plot

```python
x = [10,20,30,50,60,65,]
y = [2, 4, 1, 5, 6, 7]

plt.scatter(x,y, color = 'red', label = 'sc1')

x2 = [10, 20, 30, 40, 50]
y2 = [10, 18, 28, 38, 48]
plt.scatter(x2,y2 , color ='green', s = 50, label = 'sc2')
plt.legend()
plt.show()
```
