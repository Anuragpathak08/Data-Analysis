## Subplots in Matplotlib

**Subplots** allow us to create multiple plots within a single figure. The plots are arranged in a **grid layout**, making it possible to compare multiple visualizations within the same figure.

The `subplot()` function is used to create subplots.

### Syntax

```python
plt.subplot(rows, columns, index)
````

Where:

* `rows` → Specifies the number of rows in the grid.
* `columns` → Specifies the number of columns in the grid.
* `index` → Specifies the position of the current plot. The index starts from `1`.

### Example

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

# First subplot
plt.subplot(1, 3, 1)
plt.plot(x, [2, 4, 6, 8, 10], color='r', label='Linear')
plt.legend()

# Second subplot
plt.subplot(1, 3, 2)
plt.plot(x, [1, 4, 9, 16, 25], color='g', label='Quadratic')
plt.legend()

# Third subplot
plt.subplot(1, 3, 3)
plt.plot(x, [1, 8, 27, 64, 125], color='b', label='Cubic')
plt.legend()

plt.show()
```

### Understanding `plt.subplot(1, 3, index)`

```python
plt.subplot(1, 3, 1)
```

* `1` → One row
* `3` → Three columns
* `1` → First plot

The layout is:

```text
+-------------+-------------+-------------+
|   Plot 1    |   Plot 2    |   Plot 3    |
+-------------+-------------+-------------+
```

For example:

```python
plt.subplot(2, 2, 1)
```

creates a **2 × 2 grid**, where the plot is placed in position `1`.

```text
+-------------+-------------+
|   Plot 1    |   Plot 2    |
+-------------+-------------+
|   Plot 3    |   Plot 4    |
+-------------+-------------+
```

The `index` determines the position of each subplot and starts from `1`.


`````python

x = [1, 2, 3, 4, 5]

# First subplot
plt.subplot(1, 3, 1)
plt.plot(x, [2, 4, 6, 8, 10], color='r', label='Linear')
plt.legend()

# Second subplot
plt.subplot(1, 3, 2)
plt.plot(x, [1, 4, 9, 16, 25], color='g', label='Quadratic')
plt.legend()

# Third subplot
plt.subplot(1, 3, 3)
plt.plot(x, [1, 8, 27, 64, 125], color='b', label='Cubic')
plt.legend()

plt.show()
`````
