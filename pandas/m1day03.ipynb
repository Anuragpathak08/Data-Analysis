# DataFrame <hr>
A Pandas DataFrame is a two-dimensional, tabular data structure in Python that organizes data into rows and columns.
* It's a collection of rows and columns 
* we can also say it's a collection of Pandas Series

```python
import pandas as pd
```

`syntax :- pd.DataFrame(collection)` 

```python
pd.DataFrame([[1,2,34,5],
              [1,'hello',45,6],
              [1,23,4,5]])
```

**DF with Tuple**
<hr>

```python
pd.DataFrame(((23,45), (14,5)))
```

**DF with set**
<hr>

```python
pd.DataFrame({(2,1),(2,2),(4,5)})
```

**DF with dict**
<hr>

```python
marks = {'eng':[30,40,34,45],
         'maths':[50,0, 45, 32],
         'science':[30,40,25,46]}

pd.DataFrame(marks)
```

When initializing a DataFrame with and unequal lengths element, Pandas automatically fill the missing values with NaN.

```python
pd.DataFrame([[1,2],
              [2,3,4],
              [3,2]])
```

When creating a DataFrame, the length of the index must match the total number of rows, and the length of the columns list must match the number of columns in the data.

```python
pd.DataFrame([[1,2,2,34],
              [2,3,4,123],
              [3,2,23]],
            index=['first row', 'second row', 'third row'],
            columns=['1st', '2nd', '3rd', '4th'])
```
