# **Pandas**<hr>
**Pandas** is an open-source Python library used for data manipulation, analysis, and cleaning. Built on top of the NumPy library, it provides highly efficient, fast, and flexible data structures designed to work seamlessly with structured, tabular data—similar to SQL tables or Excel spreadsheets.

### **Core Data Structures**<hr>
1. Series: A 1-dimensional labeled array capable of holding any data type (similar to a single column or a labeled list).
2. DataFrame: A 2-dimensional labeled data structure with columns of potentially different types (similar to an Excel spreadsheet or a SQL table).

## **Series**<hr>
A Pandas Series is a one-dimensional labeled array capable of holding data of any type (integers, strings, floats, etc.). 
* It is similar to a single column in a spreadsheet or a Python list, but with an explicitly defined index that provides additional functionality for data manipulation and analysis

| Feature |  NumPy Array | Pandas Series |
| --- | --- | --- |
| **Index** | Implicit integer index only (`0, 1, 2...`) | Explicit labeled index (can be numbers, strings, or dates) |
| **Data Types** | NumPy arrays require a single homogeneous type | Homogeneous data type per Series, with specialized types (Category, Datetime) |
| **Statistical & Analysis Tools** | Basic mathematical operations (NumPy) | Rich built-in analytical methods (`.describe()`, `.value_counts()`) |
| **Memory & Performance** | NumPy is faster and uses less memory for raw math | Adds slight memory overhead to support indexes, labels, and metadata |
