### Slicing <hr>
`syntax : variable[start: end: step]`

```python
ipl[0:10:1]
# ipl[0:100:6, 'dates':'venu':] #Error because of we cannot do both rows and columns 
ipl[['date', 'team1', 'team2', 'winner']][:40:]
```

* #### **loc** <hr>

```python
america.loc[0:10:1, 'state':'zipcode':]
america.loc['state':'zipcode':] # no rows and all columns
```

```python
# ipl.loc[] # error because no slicing mentioned
ipl.loc[1:10: , 'date'::]
```

```python
df = pd.DataFrame(japan.loc[::, 'total_births'::])
df.index = japan['year']
df
```

```python
df.loc[2026:2000,]
df.loc[0:10:, 'total_births':'crude_birth_rate']
df.loc[2020.3: , 'total_births':'crude_birth_rate']
```
