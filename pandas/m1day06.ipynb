## Boolean Filtering <hr>
`syntax : variable[ condtion ]`

```python
covid[covid['Country/Region'] == 'India']

# covid[covid['Country/Region']] # error coz no condtion  and it is a syntax of indexing

covid[covid['Deaths'] > 33000]

covid[(covid['Deaths'] > 10000) & (covid['New deaths'] > 20)]

covid[(covid['New cases'] > 1000) & (covid['Deaths'] > 10000) ]['Country/Region']
```

```python
gone_case[gone_case['smokes_per_day'] > 18 ]

gone_case[(gone_case['smokes_per_day'] > 15) & (gone_case['gender'] == 'Male' ) ]

gone_case[(gone_case['smokes_per_day'] > 15) & (gone_case['gender'] == 'Male' ) ]['id'].count()

gone_case[(gone_case['smokes_per_day'] > 15) & (gone_case['gender'] == 'Female')]['id'].count()

# gone_case[(covid['Deaths'] > 15) & (gone_case['gender'] == 'Female')]['id'] # works but not prefered

# gone_case[ np.int32(10) ]  # this is indexing
```
